```python
resultados= []
dfs = [df16, df17, df18, df19, df20, df21, df22, df23, df24]
anos = list(range(2016, 2025))  # de 2016 a 2024
```

```python
for df, ano in zip(dfs, anos):
    df = df[df['Tipo do Estabelecimento'] == 'Cela física'].copy()

    for col_group in [col_medicos_fisica, col_medicos_mental, col_enfs, col_assits, col_masc, col_fem]:
        for col in col_group:
            df[col] = pd.to_numeric(df[col], errors='coerce')

    # Soma populacional por sexo
    df['pop_masc'] = df[col_masc].sum(axis=1)
    df['pop_fem'] = df[col_fem].sum(axis=1)
    df['pop_total'] = df['pop_masc'] + df['pop_fem']
    
    df['total_medicos_fisica'] = df[col_medicos_fisica].sum(axis=1)
    df['total_medicos_mental'] = df[col_medicos_mental].sum(axis=1)
    df['total_enfs'] = df[col_enfs].sum(axis=1)
    df['total_assistentes_sociais'] = df[col_assits].sum(axis=1)
    df['ano'] = ano

    grupo = df.groupby(['UF', 'ano'])[
        ['total_medicos_fisica', 'total_medicos_mental', 'total_enfs', 'total_assistentes_sociais',
         'pop_masc', 'pop_fem', 'pop_total']
    ].sum().reset_index()

    resultados.append(grupo)

# DataFrame final unificado
saude_area_populacao_ano_uf = pd.concat(resultados, ignore_index=True)
```

```python
client = bigquery.Client()
dataset_ref = client.dataset('perfil_remuneracao')
```

```python

schema=[bigquery.SchemaField('ano','INTEGER',description='Ano de referencia'),
 bigquery.SchemaField('sigla_uf','STRING',description='Sigla da Unidade da Federação'),
 bigquery.SchemaField('profissao','STRING',description='Profissional'),
 bigquery.SchemaField('total_enfs','FLOAT',description='Quantidade de enfermeiros, tecnicos e auxiliares da enfermagem'),
 bigquery.SchemaField('pop_total','FLOAT',description='Total da população prisional'),
 bigquery.SchemaField('total_enfs_por_1000','FLOAT',description='Taxa de profissionais da enfermagem por 1000 presos')
 ]
```

```python
table_ref = dataset_ref.table('SENAPPEN_profissionais_enfermagem_mil_presos') 
job_config = bigquery.LoadJobConfig(schema=schema)
job = client.load_table_from_dataframe(total_enfs_pop, table_ref, job_config=job_config)
job.result() 
```

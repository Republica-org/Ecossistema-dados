```python
dfs = [df25]
anos = list(range(2025,2026))  # de 2016 a 2024
```

```python
resultados = []
```

```python
col_enfs = ['3.1 Quantidade de Servidores que atuam no Sistema Prisional | Enfermeiros | Total',
             '3.1 Quantidade de Servidores que atuam no Sistema Prisional | Auxiliar e técnico de enfermagem | Total']

col_medicos_fisica = ['3.1 Quantidade de Servidores que atuam no Sistema Prisional | Médicos - clínicos gerais | Total',
                        '3.1 Quantidade de Servidores que atuam no Sistema Prisional | Médicos – ginecologistas | Total',
                       '3.1 Quantidade de Servidores que atuam no Sistema Prisional | Médicos - outras especialidades | Total']

col_medicos_mental = ['3.1 Quantidade de Servidores que atuam no Sistema Prisional | Psicólogos | Total',
                    '3.1 Quantidade de Servidores que atuam no Sistema Prisional | Médicos – psiquiatras | Total',
                    '3.1 Quantidade de Servidores que atuam no Sistema Prisional | Terapeuta/ terapeuta ocupacional | Total']
col_assits = ['3.1 Quantidade de Servidores que atuam no Sistema Prisional | Assistentes sociais | Total']

col_pop = ['4.1 População prisional | Total']
```

```python
for df, ano in zip(dfs, anos):
    # Filtrando apenas celas físicas
    df = df[df['Tipo do Estabelecimento'] == 'Cela física'].copy()  # .copy() evita avisos de atribuição encadeada

    # Conversão para numérico
    for col_group in [col_medicos_fisica, col_medicos_mental, col_enfs, col_assits, col_masc, col_fem]:
        for col in col_group:
            df.loc[:, col] = pd.to_numeric(df[col], errors='coerce')

    # Cálculo de totais por linha com .loc
    df.loc[:, 'total_medicos_fisica'] = df[col_medicos_fisica].sum(axis=1)
    df.loc[:, 'total_medicos_mental'] = df[col_medicos_mental].sum(axis=1)
    df.loc[:, 'total_enfs'] = df[col_enfs].sum(axis=1)
    df.loc[:, 'total_assits'] = df[col_assits].sum(axis=1)
    df.loc[:, 'pop_masc'] = df[col_masc].sum(axis=1)
    df.loc[:, 'pop_fem'] = df[col_fem].sum(axis=1)
    df.loc[:, 'pop_total'] = df[col_pop]
    df.loc[:, 'ano'] = ano

    # Agrupamento por UF e ano
    grupo = df.groupby(['UF', 'ano'])[
        ['total_medicos_fisica', 'total_medicos_mental', 'total_enfs', 'total_assits',
         'pop_masc', 'pop_fem', 'pop_total']
    ].sum().reset_index()

    resultados.append(grupo)

# Concatenação dos resultados
saude_area_populacao_ano_uf = pd.concat(resultados, ignore_index=True)
```

# Subindo no GBQ

```python
# Define the BigQuery table schema with Portuguese descriptions
schema=[bigquery.SchemaField('ano','INTEGER',description='Ano de referencia'),
        bigquery.SchemaField('sigla_uf','STRING',description='Sigla da Unidade da Federação'),
        bigquery.SchemaField('profissao','STRING',description='Profissional'),
        bigquery.SchemaField('total_medicos_mental','FLOAT',description='Quantidade de médicos responsáveis pela saúde mental'),
        bigquery.SchemaField('pop_total','FLOAT',description='Total da população prisional'),
        bigquery.SchemaField('medicos_mental_por_1000','FLOAT',description='Taxa de médicos responsáveis pela saúde mental por 1000 presos'),
        bigquery.SchemaField('medicos_mental_por_1000_1','FLOAT',description='Taxa de médicos  responsáveis pela saúde mental por 1000 presos arredondado')
 ]


# Initialize BigQuery client connection
client = bigquery.Client(project='repositoriodedadosgpsp')

# Create reference to target dataset
dataset_ref = client.dataset('perfil_remuneracao')

# Create reference to target table with standardized naming convention:
# FONTE_algo_intuitivo_dado (INEP_esforco_docente_temporal_v2)
table_ref = dataset_ref.table('SENAPPEN_profissionais_saude_mental_mil_presos_v1')

# Configure the load job with our schema definition
job_config = bigquery.LoadJobConfig(schema=schema)

# Execute the load job to upload DataFrame to BigQuery
job = client.load_table_from_dataframe(
    dataframe=df_final,
    destination=table_ref,
    job_config=job_config
)

# Wait for the job to complete
job.result()
```

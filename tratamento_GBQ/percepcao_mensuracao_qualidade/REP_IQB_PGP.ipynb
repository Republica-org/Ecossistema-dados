Subindo para o GBQ

```python
client = bigquery.Client()
dataset_ref = client.dataset('percepcao_mensuracao_qualidade_estado')
```

```python
schema= [bigquery.SchemaField('uf','STRING',description='Nome da unidade da federação'),
          bigquery.SchemaField('sigla_uf','STRING',description='Sigla do unidade da federação'),
          bigquery.SchemaField('pgp','STRING', description='Se o estado realizou a pesquisa Panorama de Gestão de Pessoas'),
          bigquery.SchemaField('iqb','STRING', description='Se o estado realizou a pesquisa Índice de Qualidade da Burocracia')]
```

```python
table_ref = dataset_ref.table('REP_igp_iqb') 
job_config = bigquery.LoadJobConfig(schema=schema)
job = client.load_table_from_dataframe(df, table_ref, job_config=job_config)
job.result()
```
